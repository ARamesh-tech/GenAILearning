```python
from langchain_ollama import ChatOllama
```

```python
llm=ChatOllama(
    model="qwen3:14b",
    temperature=0
)
```

```python
prompts = [
    {"role":"user", "content":"Hello, My name is  A.Ramesh kumaran."},
    {"role":"ai", "content":'Hello, A.Ramesh kumaran! How can I assist you today?'},
    {"role":"user", "content":"What is My name ?"}
]

res = llm.invoke(prompts)
```

```python
res.content
```

## Manage History in Langchain - for each chat window wrt chat_id

```python
history = [] # In memory history: [{"role":"user", "content":"Who is pm of india ?"}]
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit", "bye"]:
        print("Good Bye 👋")
        break

    history.append({"role":"user", "content":query})
    print("User: ", query)


    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content})
    print("AI: ", res.content, "\n")
```
